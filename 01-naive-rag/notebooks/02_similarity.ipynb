## Objective
Turn "these embeddings look close" into an actual number, and build intuition for what similarity scores mean.

See [`../theory/cosine_similarity.md`](../theory/cosine_similarity.md) for the formula and reasoning.


```python
import sys
from pathlib import Path

# Make the local `src` package importable from notebooks/
sys.path.insert(0, str(Path.cwd().parent))
```

```python
from src.embed import get_embedder
from src.retrieve import _cosine_similarity
import numpy as np

embedder = get_embedder()
```

## Compare a related pair vs. an unrelated pair


```python
pairs = [
    ("What is the refund period?", "How many days to return an item?"),
    ("What is the refund period?", "What is your favorite cake recipe?"),
    ("probation period for new hires", "90-day trial period for employees"),
]

for a, b in pairs:
    va = np.array(embedder.embed_one(a))
    vb = np.array(embedder.embed_one(b))
    score = _cosine_similarity(va.reshape(1, -1), vb)[0]
    print(f"{score:.3f}  {a!r}  <->  {b!r}")
```
