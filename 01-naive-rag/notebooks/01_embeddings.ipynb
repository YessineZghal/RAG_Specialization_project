## Objective
Generate real embeddings with the pipeline's `Embedder` abstraction and see what a vector actually looks like.

Read [`../theory/embeddings.md`](../theory/embeddings.md) first if you haven't — this notebook is the hands-on companion to it.

**Requirements:** `ollama serve` running, and `ollama pull nomic-embed-text` already run once.


```python
import sys
from pathlib import Path

# Make the local `src` package importable from notebooks/
sys.path.insert(0, str(Path.cwd().parent))
```

```python
from src.embed import get_embedder

# Default backend is Ollama (src/config.py -> EMBEDDING_BACKEND).
# This is the only cell in this notebook that talks to a local server —
# nothing is downloaded except the Ollama model, and only if you haven't
# pulled it yet (`ollama pull nomic-embed-text`).
embedder = get_embedder()
print(f"Using embedder: {type(embedder).__name__} (model={embedder.model})")
```

## Embed a few example sentences


```python
sentences = [
    "What is the refund period?",
    "How many days do I have to return an item?",
    "What is your favorite chocolate cake recipe?",
]

vectors = embedder.embed(sentences)

for sentence, vector in zip(sentences, vectors):
    print(f"{sentence!r}")
    print(f"  dim={len(vector)}  first 5 values={vector[:5]}")
```
