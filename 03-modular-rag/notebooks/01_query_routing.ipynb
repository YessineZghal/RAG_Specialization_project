## Objective
Compare rule-based and LLM-based routing across all five backends this level supports: documents (a real PDF), SQL (a real database), graph, web, and API.


```python
import sys
from pathlib import Path

LEVEL_DIR = Path.cwd().parent
for sub in ["", "routing", "multi-retriever", "sql-rag", "graph-rag", "web-rag", "api-rag", "multimodal-rag"]:
    sys.path.insert(0, str(LEVEL_DIR / sub) if sub else str(LEVEL_DIR))
```

```python
from rule_router import rule_route
from llm_router import llm_route

questions = [
    "What is the Transformer architecture based on?",
    "How many tracks are in the database?",
    "Who is affiliated with Google Brain?",
    "What is the latest news on large language models?",
    "When was the Attention Is All You Need paper published?",
    "What are the main components of multi-head attention?",
]

print(f"{'question':<55}{'rule':<12}{'llm':<12}")
for q in questions:
    rule = rule_route(q)
    llm_result = llm_route(q)
    flag = "" if rule == llm_result else "  <-- disagree"
    print(f"{q[:52]:<55}{rule:<12}{llm_result:<12}{flag}")
```

## What I observed

The two routers disagreed on 2 of 6 questions — and in this run, the **rule router was right both times, the LLM router was wrong both times**: it classified "What is the Transformer architecture based on?" as `sql`, and "What are the main components of multi-head attention?" as `api`. Both are clearly document questions about the paper's content.

This is a useful, humbling result, not a contradiction — LLM routers can mis-fire on questions that *sound* like they could reference a structured lookup ("components", "based on") even when a human reading the question wouldn't be confused. It's exactly why a real system shouldn't blindly trust either router alone: run the cheap rule router first, and reserve the LLM router (or a disagreement-triggered escalation) for cases the keyword patterns generally can't resolve at all, rather than assuming "LLM = smarter = always better."
