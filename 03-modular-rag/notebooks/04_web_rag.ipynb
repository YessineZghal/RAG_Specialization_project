# 04 — Web RAG (and a look at API RAG)

*Level 3 — Modular RAG*

## Objective
Answer a question a static PDF can never answer — something current — using **live** web search (DuckDuckGo, no API key) and real page extraction. Then contrast it with API RAG: calling a structured external service (arXiv) instead of free-text search.


```python
import sys
from pathlib import Path

LEVEL_DIR = Path.cwd().parent
for sub in ["", "web-rag", "api-rag"]:
    sys.path.insert(0, str(LEVEL_DIR / sub) if sub else str(LEVEL_DIR))
```

```python
from search import web_search
from page_extraction import search_and_extract

results = web_search("transformer neural network architecture", max_results=3)
for r in results:
    print(r["title"], "|", r["url"])
```

## Fetch and extract real page content


```python
enriched = search_and_extract("transformer neural network attention mechanism", max_results=2)
for r in enriched:
    print(f"{r['title']}")
    print(f"  {r['text'][:200]}..." if r.get('text') else f"  (fetch failed: {r.get('error')})")
    print()
```

## API RAG — a structured alternative to free-text web search

arXiv's public API returns structured metadata directly — no HTML to parse, no risk of an irrelevant page in the results.


```python
from tool_api import lookup_paper_by_id

paper = lookup_paper_by_id("1706.03762")
print("Title:", paper["title"])
print("Authors:", ", ".join(paper["authors"]))
print("Published:", paper["published"])
print("Summary:", paper["summary"][:300])
```

## What I observed

Web search is the right tool for genuinely open-ended, time-sensitive questions — but it's noisy: page extraction can fail (dead links, JS-rendered pages with no server-side text), and irrelevant results do slip through a broad query (see this level's README for a real example where a search on "latest LLM news" pulled in an unrelated result). API RAG trades that breadth for precision: when a *structured* source exists for the question you're actually asking (a paper's metadata, a stock price, a weather forecast), call it directly instead of searching for it.
